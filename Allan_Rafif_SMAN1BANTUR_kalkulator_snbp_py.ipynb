```python
def main():
    mapel_ipa = [
        "Matematika TL",
        "Matematika wajib",
        "Fisika",
        "Kimia",
        "Biologi",
        "Bahasa Indonesia",
        "Bahasa Inggris"
    ]

    print("=" * 60)
    print("  KALKULATOR RAPORT SNBP KELAS IPA  ")
    print("=" * 60)

    # 1. Input Nilai Per Mata Pelajaran (Semester 1-4)
    data_nilai = {mapel: [] for mapel in mapel_ipa}

    for mapel in mapel_ipa:
        print(f"\n--- [ Input Nilai: {mapel} ] ---")
        for sem in range(1, 5):
            while True:
                try:
                    val = float(input(f"  Semester {sem} (0-100): "))
                    if 0 <= val <= 100:
                        data_nilai[mapel].append(val)
                        break
                    else:
                        print("  Nilai harus di antara 0-100.")
                except ValueError:
                    print("  Input tidak valid. Masukkan angka.")

    # 2. Ringkasan Rata-Rata Per Mata Pelajaran
    print("\n" + "=" * 60)
    print("--- [ RINGKASAN RATA-RATA SEMESTER 1 - 4 ] ---")

    rata_mapel = {}
    total_kumulatif = 0

    for mapel, nilai_list in data_nilai.items():
        avg = sum(nilai_list) / 4
        rata_mapel[mapel] = avg
        total_kumulatif += avg
        print(f" • {mapel:<20} : {avg:.2f}")

    rata_keseluruhan = total_kumulatif / len(mapel_ipa)
    print("-" * 60)
    print(f" Rata-Rata Keseluruhan IPA (Sem 1-4) : {rata_keseluruhan:.2f}")
    print("=" * 60)

    # 3. Input Target Rata-Rata Akhir SNBP
    print("\n--- [ TARGET SEMESTER 5 ] ---")
    while True:
        try:
            target_akhir = float(input("Masukkan target rata-rata akhir (Sem 1-5) yang diinginkan: "))
            if 0 <= target_akhir <= 100:
                break
            else:
                print(" Target harus di antara 0-100.")
        except ValueError:
            print(" Input tidak valid. Masukkan angka.")

    # 4. Hitung Target Nilai Minimal Semester 5 Per Mata Pelajaran
    print("\n" + "=" * 60)
    print("--- [ REKOMENDASI TARGET NILAI SEMESTER 5 PER MAPEL ] ---")

    # Asumsi target kenaikan terdistribusi merata ke semua mapel
    for mapel, avg_saat_ini in rata_mapel.items():
        # Rumus: (Target * 5) - Total Nilai Sem 1-4
        target_sem5 = (target_akhir * 5) - sum(data_nilai[mapel])

        if target_sem5 > 100:
            status = "SANGAT TINGGI (Perlu > 100)"
        elif target_sem5 < 0:
            status = "AMAN"
            target_sem5 = 0.0
        else:
            status = "BISA DICAPAI"

        print(f" • {mapel:<20} : Target Sem 5 = {target_sem5:.2f} [{status}]")

    print("=" * 60)
    print("\nCatatan: Fokuskan peningkatan nilai pada 2 mata pelajaran pendukung")
    print("jurusan impianmu di SNBP nanti!")

if __name__ == "__main__":
    main()
```
